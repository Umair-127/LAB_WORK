Ques 1 - Read a dataset and create different line and bar plots using Matplotlib.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "hours_study": [1,2,3,4,5,6,7,8],
    "attendance": [50,60,65,70,75,80,85,90]
})

plt.plot(df["hours_study"], df["attendance"])
plt.xlabel("Hours Study")
plt.ylabel("Attendance")
plt.title("Line Plot")
plt.show()

plt.bar(df["hours_study"], df["attendance"])
plt.xlabel("Hours Study")
plt.ylabel("Attendance")
plt.title("Bar Plot")
plt.show()
```

Ques 2 - Use the same dataset and plot histograms and boxplots using Seaborn.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "hours_study": [1,2,3,4,5,6,7,8],
    "attendance": [50,60,65,70,75,80,85,90]
})

sns.histplot(df["attendance"])
plt.title("Histogram")
plt.show()

sns.boxplot(x=df["attendance"])
plt.title("Boxplot")
plt.show()
```

Ques 3 -Create scatter plots for feature relationships and generate a correlation heatmap

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "hours_study": [1,2,3,4,5,6,7,8],
    "attendance": [50,60,65,70,75,80,85,90],
    "sleep_hours": [5,6,6,7,6,7,7,8]
})

sns.scatterplot(x=df["hours_study"], y=df["attendance"])
plt.title("Scatter Plot")
plt.show()

corr = df.corr()

sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
